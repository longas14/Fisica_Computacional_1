$\large{\textbf{Cálculo del Volumen de una Esfera}} \\$

En este notebook le pedimos al usiario/a que ingrese el radio $R$ en metros de una esfera. 
Con ese dato calcula el volumen: $V = \dfrac{4\pi R^3}{3}$ y el área $A = 4\pi R^3$

```python
import math #importar librerias matemáticas

print("=== CÁLCULO DEL VOLUMEN DE UNA ESFERA ===")

# Solicitar el radio en metros 
# Usamos float() porque el radio puede ser un número decimal (e.g., 2.5)
radio = float(input("Ingresa el radio de la esfera R (en metros):\n"))
print(f"\nRadio ingresado: {radio} m")
```

$V = \dfrac{4\pi R^3}{3}$ y el área $A = 4\pi R^3$

```python
V = 4./3 * math.pi*  radio**3 # calculamos el volumen
A = 4 * math.pi*  radio**2 # calculamos el area
print("-------------------------------")
print(f"El volumen de la esfera es: {V:.4f} m^3")
print(f"En notación científica:     {V:.3e} m^3")
print(f"El area de la esfera es: {A:.4f} m^2")
print(f"En notación científica:     {A:.3e} m^2")

print("-------------------------------")
```
