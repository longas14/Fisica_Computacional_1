```python
# ==========================================================
# OSCILADOR ARMÓNICO SIMPLE
# ==========================================================

# Importar librerías
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# a) FUNCIÓN DEL OSCILADOR
# ==========================================================

def oscilador(A, m, k, phi0, t_max, num_puntos):

    # Frecuencia angular
    omega = np.sqrt(k / m)

    # Vector de tiempo
    tiempo = np.linspace(0, t_max, num_puntos)

    # Posición
    posicion = A * np.sin(omega * tiempo + phi0)

    # Velocidad
    velocidad = A * omega * np.cos(omega * tiempo + phi0)

    return tiempo, posicion, velocidad


# ==========================================================
# b) FUNCIONES DE ENERGÍA
# ==========================================================

def energia_cinetica(v, m):
    return 0.5 * m * v**2


def energia_potencial(x, k):
    return 0.5 * k * x**2


def energia_total(m, k, x, v):

    Ek = energia_cinetica(v, m)
    Ep = energia_potencial(x, k)

    return Ek + Ep


# ==========================================================
# c) PARÁMETROS DEL SISTEMA
# ==========================================================

A = 1.0          # Amplitud (m)
m = 1.0          # Masa (kg)
k = 1.0          # Constante del resorte (N/m)
phi0 = 0.0       # Fase inicial (rad)

t_max = 20       # Tiempo máximo (s)
num_puntos = 1000


# ==========================================================
# SIMULACIÓN
# ==========================================================

tiempo, posicion, velocidad = oscilador(
    A, m, k, phi0, t_max, num_puntos
)

# Energías
Ek = energia_cinetica(velocidad, m)
Ep = energia_potencial(posicion, k)
Etotal = energia_total(m, k, posicion, velocidad)


# ==========================================================
# PRIMERA FIGURA: 3 SUBGRÁFICAS
# ==========================================================

plt.figure(figsize=(10, 12))

# ----------------------------------------------------------
# Posición y velocidad
# ----------------------------------------------------------
plt.subplot(3, 1, 1)

plt.plot(tiempo, posicion, label='Posición x(t)')
plt.plot(tiempo, velocidad, label='Velocidad v(t)')

plt.xlabel('Tiempo (s)')
plt.ylabel('x (m) y v (m/s)')
plt.title('Posición y velocidad del oscilador armónico')

plt.grid(True)
plt.legend()

# ----------------------------------------------------------
# Energías cinética y potencial
# ----------------------------------------------------------
plt.subplot(3, 1, 2)

plt.plot(tiempo, Ek, label='Energía cinética')
plt.plot(tiempo, Ep, label='Energía potencial')

plt.xlabel('Tiempo (s)')
plt.ylabel('Energía (J)')
plt.title('Energías cinética y potencial')

plt.grid(True)
plt.legend()

# ----------------------------------------------------------
# Energía total
# ----------------------------------------------------------
plt.subplot(3, 1, 3)

plt.plot(tiempo, Etotal, label='Energía total')

plt.xlabel('Tiempo (s)')
plt.ylabel('Energía (J)')
plt.title('Conservación de la energía mecánica')

plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()


# ==========================================================
# COMPROBAR CONSERVACIÓN DE ENERGÍA
# ==========================================================

print("Energía mínima =", np.min(Etotal), "J")
print("Energía máxima =", np.max(Etotal), "J")
print("Energía teórica =", 0.5 * k * A**2, "J")


```
